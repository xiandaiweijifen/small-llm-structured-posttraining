# 29 External Component Verifier Suite

Run a narrow component-verifier postprocess suite on top of the Stage 14 best external prediction set.


```python
from pathlib import Path
import subprocess
import sys

def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'scripts').exists() and (candidate / 'docs').exists():
            return candidate
    raise RuntimeError('Could not find project root')

project_root = find_project_root(Path.cwd())
print('project_root =', project_root)
```

```python
completed = subprocess.run(
    [
        sys.executable,
        str(project_root / 'scripts' / 'run_external_component_verifier_suite.py'),
    ],
    check=True,
)
completed
```
