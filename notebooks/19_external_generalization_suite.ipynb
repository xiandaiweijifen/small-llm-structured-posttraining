# 19 External Generalization Suite

Run cross-dataset generalization references on a new project-format reduced-schema eval set.

Expected input format:
- project JSONL records
- each record should include `sample_id`, `input_text`, and reduced-schema `target_json`
- `schema_name` and `complexity_bucket` are optional; defaults will be filled if missing


```python
from pathlib import Path
import json

project_root = Path.cwd()
while project_root != project_root.parent and not (project_root / 'scripts').exists():
    project_root = project_root.parent

project_root
```

```python
INPUT_DATASET = project_root / 'data' / 'external_generalization' / 'new_eval_reduced.jsonl'
DATASET_TAG = 'new_eval'

print('project_root =', project_root)
print('input_dataset =', INPUT_DATASET)
print('dataset_tag =', DATASET_TAG)
print('input_exists =', INPUT_DATASET.exists())

if INPUT_DATASET.exists():
    with INPUT_DATASET.open('r', encoding='utf-8') as f:
        first_line = f.readline().strip()
    if first_line:
        sample = json.loads(first_line)
        print('sample_keys =', sorted(sample.keys()))
        print('has_target_json =', 'target_json' in sample)
        if 'target_json' in sample and isinstance(sample['target_json'], dict):
            print('target_keys =', sorted(sample['target_json'].keys()))
```

```python
import subprocess
import sys

completed = subprocess.run(
    [
        sys.executable,
        str(project_root / 'scripts' / 'run_external_generalization_reference_suite.py'),
        '--input', str(INPUT_DATASET),
        '--dataset-tag', DATASET_TAG,
    ],
    check=True,
)
completed
```
