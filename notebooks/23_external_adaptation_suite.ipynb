```python
from pathlib import Path
import subprocess
import sys


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / 'configs').exists() and (candidate / 'src').exists():
            return candidate
    raise RuntimeError('Could not find project root')


project_root = find_project_root(Path.cwd())
print('project_root =', project_root)
```

```python
build_completed = subprocess.run(
    [
        sys.executable,
        str(project_root / 'scripts' / 'build_customer_support_generalization_splits.py'),
    ],
    check=True,
)
build_completed
```

```python
run_completed = subprocess.run(
    [
        sys.executable,
        str(project_root / 'scripts' / 'run_external_adaptation_suite.py'),
    ],
    check=True,
)
run_completed
```
